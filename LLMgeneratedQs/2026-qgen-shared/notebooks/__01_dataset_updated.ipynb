THEN SELECT THE ONTOLOGIES FOR THE FEW-SHOT EXAMPLES

For each row, we select the ontologies from which we will sample the questions. For each row, the selected ontology WILL NOT include the ontology in the row (held-out, used as test).

First, each ontology is selected. Then K-(available ontologies with the same question type) are sampled with replacement.

The ontologies are SHUFFLED using the seed in the filename, MEANING THAT THEY WILL NOT APPEAR IN THE SAME ORDER.

```python
import pandas as pd
import numpy as np

filename = "../dataset/qgenllm-ds-seed-123.csv"
# get the seed from the filename
seed = int(filename.split("-")[-1].split(".")[0])
print("seed:", seed)
df = pd.read_csv(filename, sep=";").reset_index(drop=True)

print("shape:", df.shape)
print("SAMPLE")
display(df.head(20))

# build dictionary: question_type -> list of ontologies that have that question type
qt2onto = {}
for qt, gdf in df.groupby("question_type"):
    ontologies = gdf['ontology'].unique().tolist()
    qt2onto[qt] = ontologies

for k, v in qt2onto.items():
    print(f"Question type: {k}")
    print(f"  Ontologies: {v}")

rng = np.random.default_rng(seed)
df["seed"] = rng.integers(0, 2**32-1, size=df.shape[0])


# prepare a list of up to MAX_FEW_SHOTS

MAX_FEW_SHOTS = 10

def select_ontos(row):
    qt = row["question_type"]
    # select other ontos by removing the ontology in the current row
    other_ontos = [o for o in qt2onto[qt] if o != row["ontology"]]
    rng = np.random.default_rng(row["seed"])
    # shuffle other_ontos
    other_ontos = rng.permutation(other_ontos)
    # now sample with replacement up to MAX_FEW_SHOTS
    remaining_ontos = rng.choice(other_ontos, size=MAX_FEW_SHOTS-len(other_ontos), replace=True)
    selected_ontos = list(other_ontos) + list(remaining_ontos)
    return selected_ontos

df["q_from_os"] = df.apply(select_ontos, axis=1)

display(df)


```

Now select the questions using the list of ontologies built in the previous step and stored in the column `q_from_os`

The selection

```python



def select_questions(dataset):
    def sample_questions(row):
        qt = row["question_type"]
        used_qs = []
        qs = []
        axioms = []
        for o in row["q_from_os"]:
            iii = (dataset.ontology == o) & (dataset.question_type == qt) & (~dataset.row_id.isin(used_qs))
            available_qs = dataset[iii]
            if available_qs.shape[0] == 0:
                assert False, f"No available questions for ontology {o} and question type {qt}"
            # sample one question randomly
            sampled_q = available_qs.sample(n=1, random_state=row["seed"])
            qs.append(sampled_q.iloc[0]["question"])
            used_qs.append(sampled_q.iloc[0]["row_id"])
            axioms.append(sampled_q.iloc[0]["axiom"])
        return axioms, qs
    dataset["sample_axioms"], dataset["sample_questions"] = zip(*dataset.apply(sample_questions, axis=1))
    return dataset

df = select_questions(df)
```

```python
display(df.loc[:, ["sample_questions", "sample_axioms"]])

iii = [100, 150, 200]
for i in iii:
    questions = df.iloc[i]["sample_questions"]
    axioms = df.iloc[i]["sample_axioms"]
    for q, a in zip(questions, axioms):
        print(f"A: {a}")
        print(f"Q: {q}")
        print("-----")
```
